```python
#i
roll_number = 1024170258
digits = [int(d) for d in str(roll_number)]
L = [d*10 for d in digits]
print(L)

#ii
L.append(100) 
print(L)
L.insert(2, 200)  
print(L)

#iii
L.remove(100)  
print(L)
L.pop(3)  
print(L)

#iv
L.sort()
print(L)
L.sort(reverse=True)
print(L)

#v
print(L[:3])  
print(L[-3:])  

#vi
avg = sum(L)/len(L)
new_list = [x for x in L if x > avg]
print(new_list)
```

```python
scores = tuple(L[:8])
print(scores)

#i
print(max(scores), scores.index(max(scores)))
print(min(scores), scores.count(min(scores)))

#ii
reversed_scores = list(scores[::-1])  # tuples are immutable, cannot reverse in place
print(reversed_scores)

#iii
user_score = int(input("Enter a score: "))
if user_score in scores:
    print(scores.index(user_score))
else:
    print("Score not present")

#iv
try:
    scores[0] = 100
except TypeError as e:
    print(e)  # tuples are immutable, unlike lists

#v
first, second, *remaining = scores
print(first, second, remaining)
```

```python
import random

#i
roll_number = 1024170258
random.seed(roll_number)
numbers = [random.randint(100, 900) for _ in range(100)]
print(numbers)

#ii
odd_numbers = [n for n in numbers if n % 2 != 0]
print(len(odd_numbers), odd_numbers)

#iii
even_numbers = [n for n in numbers if n % 2 == 0]
print(len(even_numbers), even_numbers)

#iv
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

prime_numbers = [n for n in numbers if is_prime(n)]
print(len(prime_numbers), prime_numbers)

#v
from collections import Counter
freq = Counter(numbers)
most_common_num, count = freq.most_common(1)[0]
print(most_common_num, count)
```

```python
#i
roll_number = 1024170258
digits = [int(d) for d in str(roll_number)]
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}
print(A)
print(B)

#vi
union_set = A.union(B)
print(union_set)

#vii
intersection_set = A.intersection(B)
print(intersection_set)

#viii
diff_A_B = A.difference(B)
diff_B_A = B.difference(A)
print(diff_A_B)
print(diff_B_A)
# difference() gives elements in one set but not the other, unlike symmetric_difference() which combines both unique sides.

#ix
sym_diff = A.symmetric_difference(B)
print(sym_diff)

#x
print(A.issubset(B))
print(B.issuperset(A))

#xi
X = int(input("Enter a value to discard from set A: "))
A.discard(X)
print(A)
# discard() is safer than remove() because it does not raise an error if the element is absent.
```

```python
#initial dictionary
my_dict = {
    "name": "Arjun",
    "roll_no": "1024170258",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}
print(my_dict)

#i
my_dict["location"] = my_dict.pop("city")
print(my_dict)

#ii
my_dict["cgpa"] = 8.5
print(my_dict)

#iii
my_dict["age"] += 1
print(my_dict)

#iv
dict_pop = my_dict.copy()
dict_del = my_dict.copy()
dict_pop.pop("branch")
print(dict_pop)
del dict_del["branch"]
print(dict_del)
# pop() returns the removed value, del does not.

#v
for k, v in my_dict.items():
    print(f"{k} → {v}")

#vi
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key not present")

#vii
friend_dict = {
    "name": "Rahul",
    "roll_no": "1024170260",
    "branch": "ECE",
    "age": 21,
    "city": "Delhi"
}
merged_dict = {**my_dict, **friend_dict}
print(merged_dict)
# In case of duplicate keys, values from the second dictionary override the first.

#viii
string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(string_dict)
```
