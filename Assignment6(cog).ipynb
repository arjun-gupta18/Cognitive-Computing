```python
import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter value for M: "))

x = np.linspace(-10, 10, 400)

y1 = M * x**2
y2 = M * np.sin(x)

plt.figure(figsize=(8,6))
plt.plot(x, y1, 'r--', label='y = M*x^2')
plt.plot(x, y2, 'b-', label='y = M*sin(x)')
plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import seaborn as sns

data = {
    "Subject": ["Math", "Science", "English", "History", "Computer"],
    "Score": [85, 90, 78, 88, 95]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8,6))
barplot = sns.barplot(x="Subject", y="Score", data=df, palette="Set2")

for i, score in enumerate(df["Score"]):
    barplot.text(i, score+1, str(score), ha='center')

plt.title("Scores by Subject")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(True, axis='y')
plt.show()
```

```python
np.random.seed(12345)

data = np.random.randn(50)

fig, axs = plt.subplots(2,2, figsize=(10,8))

axs[0,0].plot(np.cumsum(data), 'g-')
axs[0,0].set_title("Cumulative Sum")
axs[0,0].grid(True)

axs[0,1].scatter(range(50), data, c='r')
axs[0,1].set_title("Scatter Plot with Noise")
axs[0,1].grid(True)


axs[1,0].hist(data, bins=10, color='b', alpha=0.7)
axs[1,0].set_title("Histogram")
axs[1,0].grid(True)

axs[1,1].boxplot(data)
axs[1,1].set_title("Boxplot")
axs[1,1].grid(True)

plt.tight_layout()
plt.show()
```

```python
sales = pd.read_csv("company_sales_data.csv")

plt.figure(figsize=(8,6))
plt.plot(sales['month_number'], sales['total_profit'], marker='o')
plt.title("Total Profit per Month")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.grid(True)
plt.show()

plt.figure(figsize=(10,6))
for col in sales.columns[1:-1]:   # skip month_number and total_profit
    plt.plot(sales['month_number'], sales[col], marker='o', label=col)

plt.title("Product Sales Data")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()

sales.plot(kind="bar", figsize=(12,6))
plt.title("Company Sales Data")
plt.xlabel("Index")
plt.ylabel("Values")
plt.grid(True)
plt.show()
```
